```python
# 3.1
# == checks whether two values are equal.
# checks whether two variables refer to the same object.
x = 3
y = 3.0
print(x == y)
print(x is y)

# 3.2
# a == b is True because both lists contain the same values.
# a is b is False because they are two separate list objects.
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)

# 3.3
# == checks value
# is checks object identity
# For small integers, Python commonly reuses integer objects, so x is y may be True.
# For larger integers, the result of is can depend on how the code is created/executed. Therefore, never use is to compare ordinary numbers. Use ==.
x = 100
y = 100
print(x is y)
print(x == y)
p = 1000
q = 1000
print(p is q)
print(p == q)

# 3.4
# Numbers, strings, lists, etc. are all objects in Python, and every object has a type.
print(type(5))
print(type("hi"))
print(type([1, 2]))
```

```python
# 7.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val
print(my_abs(5))
print(my_abs(-5))
print(my_abs(0))

# 7.2
# print(my_abs("Hi"))

# 7.3
# A function that does not explicitly return a value returns None.
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)
x = print_abs(-2.7)
print(x)

# 7.4
# val = val + 1
# only changes the local variable val.
# It does not change the original variable x.
def inc_val(val):
    val = val + 1
x = 7
inc_val(x)
print(x)

# 7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)
x = 6
y = 3
swap_message(x, y)
print("Outside function:")
print("x =", x)
print("y =", y)
```
